```python
# Cell 2: imports + config
import time, copy
import numpy as np
import pandas as pd
import lightgbm as lgb
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pack_padded_sequence
from sklearn.model_selection import train_test_split
from sklearn.metrics import roc_auc_score

SEED = 42
DATA_DIR = '.'
device = 'cuda' if torch.cuda.is_available() else ('mps' if torch.backends.mps.is_available() else 'cpu')
amp_on = (device == 'cuda')
print('device:', device, 'amp:', amp_on)

LGBM_P = {
    'objective':'binary','metric':'auc','verbose':-1,'seed':SEED,
    'learning_rate':0.018,'num_leaves':44,'min_child_samples':36,
    'feature_fraction':0.737,'bagging_fraction':0.836,'bagging_freq':1,
    'lambda_l1':0.20,'lambda_l2':2.571,
}
N_BOOST = 3000
ES = 200

SEQ_LEN = 200
n_ep = 12
bs = 256
lr = 3e-3
wd = 1e-4
alpha = 0.6

T0 = time.time()
```

```python
# Cell 3: load data
DATA_DIR = 'data'
t = time.time()
g = pd.read_csv(f'{DATA_DIR}/gender.csv')
trx = pd.read_csv(f'{DATA_DIR}/trx.csv')
tr = g[g.gender.notna()].copy(); tr['gender'] = tr['gender'].astype(int)
te = g[g.gender.isna()].copy()
tr_cids = tr.cid.values
te_cids = te.cid.values
y = tr.gender.values.astype(int)
print(f'load: {time.time()-t:.1f}s  train={len(tr)}  test={len(te)}  trx={len(trx)}')
```

```python
# Cell 4: feature engineering
t = time.time()
trx = trx.sort_values(['cid','dt']).reset_index(drop=True)
pos = trx[trx.amt>0]; neg = trx[trx.amt<0]
g_ = trx.groupby('cid')
feat = pd.DataFrame(index=g_.size().index)
feat['n_trx']=g_.size(); feat['n_days']=g_.dt.nunique()
feat['dt_min']=g_.dt.min(); feat['dt_max']=g_.dt.max()
feat['active_span']=feat['dt_max']-feat['dt_min']+1
feat['trx_per_day']=feat['n_trx']/feat['active_span']
daily = trx.groupby(['cid','dt']).size().rename('c').reset_index()
dpd = daily.groupby('cid').c
feat['dpd_mean']=dpd.mean(); feat['dpd_std']=dpd.std()

a = g_.amt
feat['amt_mean']=a.mean(); feat['amt_std']=a.std(); feat['amt_med']=a.median()
feat['amt_min']=a.min(); feat['amt_max']=a.max(); feat['amt_sum']=a.sum()
feat['amt_skew']=a.skew()
feat['amt_kurt']=a.apply(lambda x: x.kurt() if len(x)>3 else 0.0)
feat['amt_q05']=a.quantile(0.05); feat['amt_q25']=a.quantile(0.25)
feat['amt_q75']=a.quantile(0.75); feat['amt_q95']=a.quantile(0.95)

abs_a = trx.amt.abs()
feat['amt_abs_mean']=abs_a.groupby(trx.cid).mean()
feat['amt_abs_sum']=abs_a.groupby(trx.cid).sum()
feat['amt_abs_max']=abs_a.groupby(trx.cid).max()
feat['concentration']=feat['amt_abs_max']/(feat['amt_abs_sum']+1)
log_a = np.log1p(abs_a)
feat['log_amt_mean']=log_a.groupby(trx.cid).mean()
feat['log_amt_std']=log_a.groupby(trx.cid).std()

pg=pos.groupby('cid').amt; ng=neg.groupby('cid').amt
feat['pos_n']=pg.size().reindex(feat.index).fillna(0)
feat['pos_mean']=pg.mean().reindex(feat.index)
feat['pos_sum']=pg.sum().reindex(feat.index).fillna(0)
feat['neg_n']=ng.size().reindex(feat.index).fillna(0)
feat['neg_mean']=ng.mean().reindex(feat.index)
feat['neg_sum']=ng.sum().reindex(feat.index).fillna(0)
feat['cd_ratio']=feat['pos_n']/(feat['neg_n']+1)
feat['cd_amt_ratio']=feat['pos_sum']/(feat['neg_sum'].abs()+1)

def _mode(x):
    v=x.value_counts(); return v.index[0] if len(v) else -1
feat['n_mcc']=g_.mcc.nunique(); feat['n_ttc']=g_.ttc.nunique(); feat['n_tid']=g_.tid.nunique()
feat['mcc_mode']=g_.mcc.agg(_mode); feat['ttc_mode']=g_.ttc.agg(_mode)
mcl = trx.groupby(['cid','mcc']).size().rename('c').reset_index().sort_values(['cid','c'], ascending=[True,False])
feat['mcc_top3_share']=mcl.groupby('cid').head(3).groupby('cid').c.sum()/feat['n_trx']

def share(mask, name):
    s = trx[mask].groupby('cid').size().reindex(feat.index).fillna(0)
    feat[name] = s/feat['n_trx']
share(trx.mcc.between(56,61),'sh_food'); share(trx.mcc.between(5,18),'sh_transport')
share(trx.mcc.between(69,79),'sh_clothing'); share(trx.mcc.between(153,165),'sh_entert')
share(trx.ttc.between(10,19),'sh_cash'); share(trx.ttc.between(25,54),'sh_c2c')
share(trx.ttc==23,'sh_online'); share(trx.ttc.isin([1,4,6]),'sh_pos')

diffs = trx.groupby('cid').dt.diff(); diffs.index = trx.cid.values
feat['gap_mean']=diffs.groupby(level=0).mean(); feat['gap_std']=diffs.groupby(level=0).std()
trx['dow']=trx.dt%7; trx['is_we']=trx.dow.isin([5,6]).astype(int)
feat['sh_weekend']=trx.groupby('cid').is_we.mean()
h = pd.to_datetime(trx.tm, format='%H:%M:%S', errors='coerce').dt.hour
feat['hour_mean']=h.groupby(trx.cid).mean()
feat['hour_std']=h.groupby(trx.cid).std()
feat['sh_night']=((h<6)|(h>=22)).groupby(trx.cid).mean()

trx['amt_abs'] = trx.amt.abs()
mcc_cnt = trx.groupby(['cid','mcc']).size().unstack(fill_value=0)
mcc_sum = trx.groupby(['cid','mcc']).amt_abs.sum().unstack(fill_value=0)
mcc_share = mcc_sum.div(mcc_sum.sum(axis=1).replace(0,1), axis=0)
probs = mcc_cnt.div(mcc_cnt.sum(axis=1).replace(0,1), axis=0)
mcc_entropy = -(probs*np.log(probs+1e-12)).sum(axis=1).rename('mcc_entropy')
mcc_piv = pd.concat([mcc_cnt.add_prefix('mcc_cnt_'), mcc_sum.add_prefix('mcc_sum_'),
                     mcc_share.add_prefix('mcc_share_'), mcc_entropy], axis=1)

top_ttc = trx.ttc.value_counts().head(30).index
tsub = trx[trx.ttc.isin(top_ttc)]
ttc_cnt = tsub.groupby(['cid','ttc']).size().unstack(fill_value=0).reindex(sorted(trx.cid.unique()), fill_value=0)
ttc_share = ttc_cnt.div(ttc_cnt.sum(axis=1).replace(0,1), axis=0)
ttc_piv = pd.concat([ttc_cnt.add_prefix('ttc_cnt_'), ttc_share.add_prefix('ttc_share_')], axis=1)

dow_hist = pd.crosstab(trx.cid, trx.dow, normalize='index')
dow_hist.columns = [f'dow_sh_{c}' for c in dow_hist.columns]

mid = (trx.dt.min()+trx.dt.max())/2
early = trx[trx.dt<=mid]; late = trx[trx.dt>mid]
cids_all = sorted(trx.cid.unique())
def _agg(df, suf):
    a=df.groupby('cid').agg(n=('amt','size'), s=('amt','sum'), nm=('mcc','nunique'))
    a.columns=[f'{c}_{suf}' for c in ['n','s','nm']]
    return a.reindex(cids_all, fill_value=0)
tmp = pd.concat([_agg(early,'early'), _agg(late,'late')], axis=1)
tmp['n_late_early_r']=tmp['n_late']/(tmp['n_early']+1)
tmp['s_late_early_r']=tmp['s_late']/(tmp['s_early'].abs()+1)
tmp['nm_late_early_r']=tmp['nm_late']/(tmp['nm_early']+1)

feat = pd.concat([feat, mcc_piv, ttc_piv, dow_hist, tmp], axis=1).reset_index()
if feat.columns[0]!='cid':
    feat = feat.rename(columns={feat.columns[0]:'cid'})

# gender priors — full-train (no OOF)
cnt_tr = mcc_cnt.loc[tr_cids].values.astype(np.float32)
cnt_te = mcc_cnt.loc[te_cids].values.astype(np.float32)
used_tr = (cnt_tr>0).astype(np.float32)
yf = y.astype(np.float32)
ucnt = used_tr.sum(axis=0)
rates = np.where(ucnt>=5, (used_tr.T @ yf)/np.maximum(ucnt,1), 0.5)
fm = rates<0.4; mm = rates>0.6
tot_tr = cnt_tr.sum(axis=1).clip(min=1); tot_te = cnt_te.sum(axis=1).clip(min=1)
tr_w = (cnt_tr*rates).sum(axis=1)/tot_tr
tr_f_ = cnt_tr[:,fm].sum(axis=1)/tot_tr
tr_m = cnt_tr[:,mm].sum(axis=1)/tot_tr
te_w_ = (cnt_te*rates).sum(axis=1)/tot_te
te_f_ = cnt_te[:,fm].sum(axis=1)/tot_te
te_m = cnt_te[:,mm].sum(axis=1)/tot_te
gp = pd.concat([
    pd.DataFrame({'cid':tr_cids,'gp_w':tr_w,'gp_fem':tr_f_,'gp_mal':tr_m}),
    pd.DataFrame({'cid':te_cids,'gp_w':te_w_,'gp_fem':te_f_,'gp_mal':te_m}),
], ignore_index=True)
feat = feat.merge(gp, on='cid', how='left')

tr_d = tr.merge(feat, on='cid', how='left')
te_d = te.merge(feat, on='cid', how='left')
TOP5 = ['gp_w','gp_fem','gp_mal','sh_clothing','mcc_share_116']
def add_ix(df, t5):
    eps=1e-6; a,b,c,d,e=t5[:5]
    df[f'ix_{a}_div_{b}']=df[a]/(df[b]+eps); df[f'ix_{a}_div_{c}']=df[a]/(df[c]+eps)
    df[f'ix_{b}_div_{c}']=df[b]/(df[c]+eps); df[f'ix_{a}_div_{d}']=df[a]/(df[d]+eps)
    df[f'ix_{a}_div_{e}']=df[a]/(df[e]+eps)
    df[f'ix_{a}_x_{b}']=df[a]*df[b]; df[f'ix_{a}_x_{c}']=df[a]*df[c]
    df[f'ix_{b}_x_{c}']=df[b]*df[c]; df[f'ix_{a}_x_{d}']=df[a]*df[d]
    df[f'ix_{a}_x_{e}']=df[a]*df[e]
    return df
tr_d = add_ix(tr_d, TOP5); te_d = add_ix(te_d, TOP5)
cols = [c for c in tr_d.columns if c not in ['cid','gender']]
X_tr = tr_d[cols]; X_te = te_d[cols]
print(f'features: {time.time()-t:.1f}s  X_tr={X_tr.shape}  X_te={X_te.shape}')
```

```python
# Cell 5: LightGBM — 80/20 holdout to find best_iter, refit on full
t = time.time()
Xt, Xv, yt, yv = train_test_split(X_tr, y, test_size=0.2, stratify=y, random_state=SEED)
m = lgb.train(LGBM_P, lgb.Dataset(Xt, yt), num_boost_round=N_BOOST,
              valid_sets=[lgb.Dataset(Xv, yv)],
              callbacks=[lgb.early_stopping(ES), lgb.log_evaluation(0)])
best_iter = m.best_iteration
ho_auc = roc_auc_score(yv, m.predict(Xv, num_iteration=best_iter))
print(f'lgbm holdout: best_iter={best_iter}  auc={ho_auc:.5f}')

m_full = lgb.train(LGBM_P, lgb.Dataset(X_tr, y), num_boost_round=best_iter)
pred_lgb = m_full.predict(X_te)
print(f'lgbm: {time.time()-t:.1f}s')
```

```python
# Cell 6: build sequences
t = time.time()
trx_s = trx.sort_values(['cid','dt']).reset_index(drop=True)
trx_s['amt_p'] = (np.sign(trx_s.amt)*np.log1p(trx_s.amt.abs())).astype(np.float32)
trx_s['dt_diff'] = trx_s.groupby('cid').dt.diff().fillna(0).astype(np.float32)
trx_s['mcc_i'] = (trx_s.mcc + 1).astype(np.int64)
trx_s['ttc_i'] = (trx_s.ttc + 1).astype(np.int64)
trx_s['dow_i'] = (trx_s.dt % 7).astype(np.int64)

all_cids = sorted(g.cid.unique())
cid_to_idx = {c:i for i,c in enumerate(all_cids)}
tail = trx_s.groupby('cid', sort=False).tail(SEQ_LEN).copy()
tail['pos'] = tail.groupby('cid').cumcount()
tail['ci'] = tail.cid.map(cid_to_idx)
n_all = len(all_cids)

MCC = np.zeros((n_all, SEQ_LEN), dtype=np.int64)
TTC = np.zeros((n_all, SEQ_LEN), dtype=np.int64)
DOW = np.zeros((n_all, SEQ_LEN), dtype=np.int64)
NUM = np.zeros((n_all, SEQ_LEN, 2), dtype=np.float32)
ci_a = tail.ci.values; po_a = tail.pos.values
MCC[ci_a, po_a] = tail.mcc_i.values
TTC[ci_a, po_a] = tail.ttc_i.values
DOW[ci_a, po_a] = tail.dow_i.values
NUM[ci_a, po_a, 0] = tail.amt_p.values
NUM[ci_a, po_a, 1] = tail.dt_diff.values
LEN = tail.groupby('ci').size().reindex(range(n_all), fill_value=0).values.astype(np.int64)
LEN = np.maximum(LEN, 1)

n_mcc_v = int(trx.mcc.max())+2
n_ttc_v = int(trx.ttc.max())+2
tr_idx = np.array([cid_to_idx[c] for c in tr_cids])
te_idx = np.array([cid_to_idx[c] for c in te_cids])
print(f'sequences: {time.time()-t:.1f}s  n_mcc={n_mcc_v}  n_ttc={n_ttc_v}')
```

```python
# Cell 7: GRU model
class SeqDS(Dataset):
    def __init__(self, mcc, ttc, dow, num, tab, lens, y=None):
        self.mcc=mcc; self.ttc=ttc; self.dow=dow; self.num=num; self.tab=tab; self.lens=lens; self.y=y
    def __len__(self): return len(self.mcc)
    def __getitem__(self, i):
        yv = float(self.y[i]) if self.y is not None else 0.0
        return (torch.from_numpy(self.mcc[i]),
                torch.from_numpy(self.ttc[i]),
                torch.from_numpy(self.dow[i]),
                torch.from_numpy(self.num[i]),
                torch.from_numpy(self.tab[i]),
                int(self.lens[i]),
                torch.tensor(yv, dtype=torch.float32))

class SeqNet(nn.Module):
    def __init__(self, n_mcc, n_ttc, tab_dim, e_dim=16, h_dim=64):
        super().__init__()
        self.mcc_e = nn.Embedding(n_mcc, e_dim, padding_idx=0)
        self.ttc_e = nn.Embedding(n_ttc, e_dim, padding_idx=0)
        self.dow_e = nn.Embedding(7, 8)
        self.num_p = nn.Linear(2, 8)
        self.gru = nn.GRU(e_dim*2+16, h_dim, num_layers=1, batch_first=True, bidirectional=True)
        self.drop1d = nn.Dropout1d(0.2)
        self.tab_bn = nn.BatchNorm1d(tab_dim)
        self.fc1 = nn.Linear(h_dim*4 + tab_dim, 128)
        self.bn1 = nn.BatchNorm1d(128)
        self.relu = nn.ReLU()
        self.drop = nn.Dropout(0.3)
        self.head = nn.Linear(128, 1)
    def forward(self, mcc, ttc, dow, num, tab, lens):
        m = self.mcc_e(mcc); tc = self.ttc_e(ttc); d = self.dow_e(dow); nm = self.num_p(num)
        x = torch.cat([m, tc, d, nm], dim=-1)
        x = self.drop1d(x.transpose(1, 2)).transpose(1, 2)
        packed = pack_padded_sequence(x, lens.cpu(), batch_first=True, enforce_sorted=False)
        out, _ = self.gru(packed)
        out, _ = torch.nn.utils.rnn.pad_packed_sequence(out, batch_first=True)
        
        mask = torch.arange(out.size(1), device=out.device)[None, :] < lens[:, None]
        mask = mask.unsqueeze(-1)
        out = out.masked_fill(~mask, 0.0)
        sum_out = out.sum(dim=1)
        mean_pool = sum_out / lens.unsqueeze(-1).clamp(min=1).float()
        
        out_max = out.masked_fill(~mask, -1e9)
        max_pool, _ = out_max.max(dim=1)
        
        t_feat = self.tab_bn(tab)
        h = torch.cat([mean_pool, max_pool, t_feat], dim=-1)
        
        h = self.fc1(h)
        h = self.bn1(h)
        h = self.relu(h)
        h = self.drop(h)
        return self.head(h).squeeze(-1)

print('model defined')
```

```python
# Cell 8: train GRU — 90/10 holdout, track best epoch state
t = time.time()
MCC_tr = MCC[tr_idx]; TTC_tr = TTC[tr_idx]; DOW_tr = DOW[tr_idx]; NUM_tr = NUM[tr_idx]; LEN_tr = LEN[tr_idx]
MCC_te = MCC[te_idx]; TTC_te = TTC[te_idx]; DOW_te = DOW[te_idx]; NUM_te = NUM[te_idx]; LEN_te = LEN[te_idx]

TAB_tr = np.nan_to_num(X_tr.values.astype(np.float32))
TAB_te = np.nan_to_num(X_te.values.astype(np.float32))

tr_i, va_i = train_test_split(np.arange(len(tr_idx)), test_size=0.10, stratify=y, random_state=SEED)
tr_ds = SeqDS(MCC_tr[tr_i], TTC_tr[tr_i], DOW_tr[tr_i], NUM_tr[tr_i], TAB_tr[tr_i], LEN_tr[tr_i], y[tr_i])
va_ds = SeqDS(MCC_tr[va_i], TTC_tr[va_i], DOW_tr[va_i], NUM_tr[va_i], TAB_tr[va_i], LEN_tr[va_i], y[va_i])
te_ds = SeqDS(MCC_te, TTC_te, DOW_te, NUM_te, TAB_te, LEN_te)
tr_loader = DataLoader(tr_ds, batch_size=bs, shuffle=True, num_workers=0)
va_loader = DataLoader(va_ds, batch_size=bs, shuffle=False, num_workers=0)
te_loader = DataLoader(te_ds, batch_size=bs, shuffle=False, num_workers=0)

net = SeqNet(n_mcc_v, n_ttc_v, tab_dim=TAB_tr.shape[1]).to(device)
opt = torch.optim.AdamW(net.parameters(), lr=lr, weight_decay=wd)
n_steps = n_ep * len(tr_loader)
sch = torch.optim.lr_scheduler.OneCycleLR(opt, max_lr=lr, total_steps=n_steps)
scaler = torch.amp.GradScaler('cuda', enabled=amp_on)
loss_fn = nn.BCEWithLogitsLoss()

best_auc = 0.0
best_state = None
for ep in range(n_ep):
    net.train(); losses=[]
    for m_b, t_b, d_b, n_b, tb_b, l_b, y_b in tr_loader:
        m_b=m_b.to(device); t_b=t_b.to(device); d_b=d_b.to(device); n_b=n_b.to(device); tb_b=tb_b.to(device); y_b=y_b.to(device)
        l_b=torch.as_tensor(l_b)
        opt.zero_grad()
        if amp_on:
            with torch.amp.autocast('cuda'):
                logit = net(m_b, t_b, d_b, n_b, tb_b, l_b); loss = loss_fn(logit, y_b)
            scaler.scale(loss).backward(); scaler.step(opt); scaler.update()
        else:
            logit = net(m_b, t_b, d_b, n_b, tb_b, l_b); loss = loss_fn(logit, y_b)
            loss.backward(); opt.step()
        sch.step(); losses.append(loss.item())
    net.eval(); ps=[]
    with torch.no_grad():
        for m_b, t_b, d_b, n_b, tb_b, l_b, _ in va_loader:
            m_b=m_b.to(device); t_b=t_b.to(device); d_b=d_b.to(device); n_b=n_b.to(device); tb_b=tb_b.to(device)
            l_b=torch.as_tensor(l_b)
            logit = net(m_b, t_b, d_b, n_b, tb_b, l_b)
            ps.append(torch.sigmoid(logit).float().cpu().numpy())
    va_pred = np.concatenate(ps)
    va_auc = roc_auc_score(y[va_i], va_pred)
    print(f'ep {ep:2d}  loss={np.mean(losses):.4f}  ho_auc={va_auc:.5f}')
    if va_auc > best_auc:
        best_auc = va_auc
        best_state = copy.deepcopy(net.state_dict())

net.load_state_dict(best_state)
print(f'best holdout auc: {best_auc:.5f}')
print(f'gru: {time.time()-t:.1f}s')
```

```python
# Cell 9: predict + blend + save
t = time.time()
net.eval(); ps=[]
with torch.no_grad():
    for m_b, t_b, d_b, n_b, tb_b, l_b, _ in te_loader:
        m_b=m_b.to(device); t_b=t_b.to(device); d_b=d_b.to(device); n_b=n_b.to(device); tb_b=tb_b.to(device)
        l_b=torch.as_tensor(l_b)
        logit = net(m_b, t_b, d_b, n_b, tb_b, l_b)
        ps.append(torch.sigmoid(logit).float().cpu().numpy())
pred_gru = np.concatenate(ps)

pred_blend = alpha * pred_lgb + (1 - alpha) * pred_gru
sub = pd.DataFrame({'cid': te_cids, 'gender': pred_blend})
sub.to_csv('submission.csv', index=False)
print(f'predict+blend: {time.time()-t:.1f}s  alpha={alpha}')
print(sub.head())
```
